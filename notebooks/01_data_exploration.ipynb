# 01 - Data Exploration

Explore synthetic borrower + financial statement data.
Goal: understand distributions, industry segmentation, and outlier patterns before spreading.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
pd.set_option("display.max_columns", None)

borrowers = pd.read_csv("../data/synthetic/borrowers.csv")
financials = pd.read_csv("../data/synthetic/financials.csv")

print("Borrowers:", borrowers.shape)
print("Financials:", financials.shape)
borrowers.head()
```

```python
# Industry distribution
borrowers["industry"].value_counts().plot(kind="bar", color="steelblue")
plt.title("Borrowers by Industry")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
# Revenue distribution by industry (using last year)
latest_year = financials["fiscal_year"].max()
rev = financials[
    (financials["standard_line_item"] == "Revenue") &
    (financials["fiscal_year"] == latest_year)
].merge(borrowers[["borrower_id", "industry"]], on="borrower_id")

plt.figure(figsize=(10, 5))
sns.boxplot(data=rev, x="industry", y="amount")
plt.xticks(rotation=45)
plt.title(f"Revenue Distribution by Industry (FY {latest_year})")
plt.tight_layout()
plt.show()
```

```python
# Coverage: how many line items per borrower-year
cov = financials.groupby(["borrower_id", "fiscal_year"]).size()
print("Line items per borrower-year:")
print(cov.describe())

cov.hist(bins=20, color="teal")
plt.title("Line-item Coverage per Borrower-Year")
plt.tight_layout()
plt.show()
```

## Observations

- 120 borrowers across 6 industries
- Revenue distribution shows industry-level separation
- Line-item coverage is consistent (~25 items per borrower-year)
- Ready for spreading pipeline
